# U.S. Medical Insurance Costs--Step3. Linear Regression Model

```python
import numpy as np
# Create a multiple linear regression model: charges = age + bmi + children + smoker + sex + region
import statsmodels.api as sm
import pandas as pd
```

```python
ins_cleaned_df = pd.read_csv("/Users/shenkong/Desktop/Projects/pj02_US_Medical_Insurance_Costs/data/ins_cleaned.csv")
```

```python
X_continuous = ins_cleaned_df[["age", "bmi", "children"]]  # Use two pairs of brackets to include [x1, x2, ... ...]
X_continuous = sm.add_constant(X_continuous)

y = ins_cleaned_df["charges"]
# X includes all the continuous independent variables, y is the dependent variable--"charges"

model = sm.OLS(y, X_continuous).fit()
model.summary()
```

```python
# Conclusion: The R-squared value is 0.120, which means the model can explain only 12% of the variation in insurance charges
# The relatively low R-squared suggests that important factors affecting insurance charges may not be included in the current model
```

```python
# Add the categorical variables to the model:
# Need to encode at first
df_encoded = pd.get_dummies(ins_cleaned_df, drop_first=True)
X1 = sm.add_constant(df_encoded[["age", "bmi", "children", "smoker_yes"]].astype(float))

model1 = sm.OLS(y, X1).fit()
model1.summary()
# The model using only the smoking status yields an R-squared of approximately 0.62 and an adjusted R-squared of approximately 0.62, indicating that smoking alone explains about 62% of the variation in insurance charges. This suggests that smoking status is the strongest predictor in the dataset.
```

```python
X2 = sm.add_constant(df_encoded[["age", "bmi", "children", "smoker_yes", "sex_male"]].astype(float))

model2 = sm.OLS(y, X2).fit()
model2.summary()
# The inclusion of gender does not improve the model’s R-squared or adjusted R-squared, indicating that sex provides little to no additional explanatory power beyond smoking status.
```

```python
X3 = sm.add_constant(df_encoded[["age", "bmi", "children", "smoker_yes", "region_northwest", "region_southwest", "region_southeast"]].astype(float))

model3 = sm.OLS(y, X3).fit()
model3.summary()
```

```python
# Summary table:
summary_table = pd.DataFrame({
                            "Variable": [
                                "age",
                                "bmi",
                                "children",
                                "smoker_yes",
                                "sex_male",
                                "region"
                            ],
                            "Contribution":[
                                "Moderate(~+8% R-squared)",
                                "Weak(~+3% R-squared)",
                                "Moderate(~+6% R-squared)",
                                "Dominant(~75%)",
                                "Negligible",
                                "Negligible"
                            ]
})
summary_table
```

```python
# Linear Model:
X = sm.add_constant(df_encoded[["age", "bmi", "children", "smoker_yes"]].astype(float))

model = sm.OLS(y, X).fit()
model.summary()
```

```python
# Output the final linear regression equation:
coef = model.params  # Extract all the coefficients

print(f"Final Linear Regression Equation: "
      f"charges = {round(coef['const'], 2)} "
      f"+ {round(coef['age'], 2)} * age "
      f"+ {round(coef['bmi'], 2)} * bmi "
      f"+ {round(coef['children'], 2)} * children "
      f"+ {round(coef['smoker_yes'], 2)} * smoker_yes")
```

```python
# Conclusion:
# The final OLS regression model explains about 75% of the variance in insurance charges, indicating a strong overall fit and substantial explanatory power for this dataset
# Smoking status is the strongest driver of insurance charges, with smokers paying significantly more even after controlling for other factors. Age, BMI, and number of children are also statistically significant but have smaller effects.
# Overall, insurance costs are mainly driven by behavioral and health factors, especially smoking, while demographic variables play a secondary role.
```

```python
# Make prediction based on the model:
y_pred = model.predict(X)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
)

mae_ols = mean_absolute_error(y, y_pred)
rmse_ols = np.sqrt(
    mean_squared_error(y, y_pred)
)

print(f"R-squared: 0.750")
print(f"MAE: {mae_ols}")
print(f"RMSE: {rmse_ols}")
```
